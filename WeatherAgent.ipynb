```python
import requests
```

# Getting geo datas

```python
url = "https://geocoding-api.open-meteo.com/v1/search"
city = input("Tell me the city you want the weather informations about: ")
print("your requested city is ", city)
params = {'name': city}
response = requests.get(url, params = params)
print(response.status_code)
if response.status_code == 200:
  data = response.json()
  latitude = data["results"][0]["latitude"]
  longitude = data["results"][0]["longitude"]
  print("Latitude:", latitude)
  print("Longitude:", longitude)
else:
  print("Error:", response.status_code)
```

# Identifying the weather status

```python
url_2 = "https://api.open-meteo.com/v1/forecast" # Corrected URL for weather forecast
params_2 = {'latitude': latitude, 'longitude': longitude,'hourly': 'weathercode,temperature_2m' }
response_2 = requests.get(url_2, params = params_2)
print(response_2.status_code)
if response_2.status_code == 200:
  data_2 = response_2.json()
  hourly = data_2["hourly"]['weathercode'][0]
  Temp = data_2["hourly"]['temperature_2m'][0]
else:
  print("Error:", response_2.status_code)
weather_codes = {
    0: "Clear sky ☀️",
    1: "Mainly clear 🌤️",
    2: "Partly cloudy ⛅",
    3: "Overcast ☁️",
    45: "Foggy 🌫️",
    48: "Icy fog 🌫️❄️",
    51: "Light drizzle 🌦️",
    53: "Moderate drizzle 🌦️",
    55: "Dense drizzle 🌦️",
    61: "Slight rain 🌧️",
    63: "Moderate rain 🌧️",
    65: "Heavy rain 🌧️",
    71: "Slight snow 🌨️",
    73: "Moderate snow 🌨️",
    75: "Heavy snow 🌨️",
    77: "Snow grains 🌨️",
    80: "Slight showers 🌦️",
    81: "Moderate showers 🌦️",
    82: "Violent showers 🌦️",
    85: "Slight snow showers 🌨️",
    86: "Heavy snow showers 🌨️",
    95: "Thunderstorm ⛈️",
    96: "Thunderstorm with slight hail ⛈️🌩️",
    99: "Thunderstorm with heavy hail ⛈️🌩️"
}
Weather_Status = weather_codes[hourly]
print(Weather_Status,"\n",Temp)
```

# Posting the Infos

```python
post_url = "https://postman-echo.com/post"
WeatherInfo = {
    'city' : city,
    'Temperature' : Temp,
    'Weather Status' :  Weather_Status
}
headers = {'Authorization' : 'Bearer YOUR_KEY'}
response_3 = requests.post(post_url, json = WeatherInfo, headers = headers)
print(response_3.status_code)
print(response_3.json())
```
